```python
import matplotlib.pyplot as plt
import numpy as np

# Ganti dengan hasil evaluasi Anda
metrik = ["Faithfulness", "Answer\nRelevancy"]
baseline = [0.78, 0.82]
fusion = [0.86, 0.89]

x = np.arange(len(metrik))
width = 0.35

plt.figure(figsize=(7, 5))
plt.bar(x - width/2, baseline, width, label="RAG Baseline")
plt.bar(x + width/2, fusion, width, label="RAG Fusion")

plt.xticks(x, metrik)
plt.ylabel("Skor")
plt.ylim(0, 1.0)
plt.title("Comparison of Generation Metrics")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
